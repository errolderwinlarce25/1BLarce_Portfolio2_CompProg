CONCEPT: SPECIATION ANALYZER

In an analytical chemistry calculation, the state of an acid (protonated/deprotonated) can be determined based on the relationship between the pH of the solution and the pKa of the chemical.

This code uses conditional statements to decide which species dominates.

```python
def check_chemical_speciation(pH, pKa):
    # Calculate the difference to determine the dominant species
    diff = pH - pKa
    
    print(f"Current pH: {pH}")
    print(f"Substance pKa: {pKa}")
    print("-" * 30)

    # Conditionals: Logic based on Analytical Chemistry Calculations
    if diff <= -1:
        # pH is at least 1 unit below pKa (mostly protonated)
        return "DOMINANT SPECIES: Acidic form [HA]. Logic: Protonated (>90%)."
    
    elif diff >= 1:
        # pH is at least 1 unit above pKa (mostly deprotonated)
        return "DOMINANT SPECIES: Basic form [A-]. Logic: Deprotonated (>90%)."
    
    else:
        # pH is near the pKa (buffer region)
        percentage_a_minus = (10**diff / (1 + 10**diff)) * 100
        return f"STATUS: Buffer Region. Mixture of forms. [A-] is approx {percentage_a_minus:.1f}%."

# Simulation: Checking Acetic Acid (pKa ~4.76) in a solution of pH 6.0
print(check_chemical_speciation(pH=6.0, pKa=4.76))
```
